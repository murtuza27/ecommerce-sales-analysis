```python
import pandas as pd
import numpy as np

data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],

    "Product": [
        "Laptop", "Phone", "Headphones", "Laptop", "Phone",
        "Keyboard", "Monitor", "Headphones", "Monitor", "Laptop"
    ],

    "Category": [
        "Electronics", "Electronics", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories",
        "Electronics", "Electronics"
    ],

    "Quantity": [1, 2, 3, 1, 1, 2, 1, 2, 2, 1],

    "Price": [
        1200, 800, 150, 1250, 820,
        100, 400, 150, 420, 1180
    ],

    "Customer": [
        "Alice", "Bob", "Charlie", "David", "Alice",
        "Bob", "Emma", "Charlie", "David", "Emma"
    ]
}

df = pd.DataFrame(data)

df["Revenue"] = df["Quantity"] * df["Price"]

print(df)
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum()

print(category_revenue)
```

```python
total_revenue = df["Revenue"].sum()

category_percentage = ((category_revenue / total_revenue) * 100).round(2)

print(category_percentage)
```
